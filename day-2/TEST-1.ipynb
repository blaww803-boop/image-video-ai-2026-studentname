```python
import cv2

image = cv2.imread(r"C:\Users\PC-LAB1\Desktop\MHB\WhatsApp Image 2026-09-22 at 10.41.38 AM.jpeg")

print("Shape:", image.shape)
print("Data type:", image.dtype)

cv2.imshow("Original Image", image)
cv2.waitKey(0)
cv2.destroyAllWindows()

#cv2.imwrite(r"C:\Users\PC-LAB1\Desktop\lenas.jpg", image)
# Resize
small = cv2.resize(image, (400, 300))

# Crop
crop = image[100:400, 150:550]

# Flip
horizontal = cv2.flip(image, 1)
vertical = cv2.flip(image, 0)

cv2.imwrite(r"C:\Users\PC-LAB1\Desktop\MHB\day-2\small.jpg", small)
cv2.imwrite(r"C:\Users\PC-LAB1\Desktop\MHB\day-2\crop.jpg", crop)
cv2.imwrite(r"C:\Users\PC-LAB1\Desktop\MHB\day-2\horizontal.jpg", horizontal)
cv2.imwrite(r"C:\Users\PC-LAB1\Desktop\MHB\day-2\vertical.jpg", vertical)
```

```python
import cv2
import matplotlib.pyplot as plt

height, width = image.shape[:2]

# مركز الصورة
center = (width // 2, height // 2)

# مصفوفة الدوران 180 درجة
matrix = cv2.getRotationMatrix2D(center, 180, 1.0)

# تدوير الصورة
rotated = cv2.warpAffine(
    image,
    matrix,
    (width, height)
)

# عرض الصورة
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

# تحويل BGR إلى BRG
brg = image[:, :, [0, 2, 1]]

print("BGR:", image.shape)
print("RGB:", rgb.shape)
print("BRG:", brg.shape)
print("Gray:", gray.shape)
print("HSV:", hsv.shape)
```

```python
import matplotlib.pyplot as plt

# BGR → RGB للعرض
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# BGR → BRG
brg = image[:, :, [0, 2, 1]]

# HSV → RGB للعرض
hsv_rgb = cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB)

plt.figure(figsize=(16, 4))

# BGR / Color
plt.subplot(1, 4, 1)
plt.imshow(rgb)
plt.title("BGR / Color")
plt.axis("off")

# BRG
plt.subplot(1, 4, 2)
plt.imshow(brg)
plt.title("BRG")
plt.axis("off")

# Gray
plt.subplot(1, 4, 3)
plt.imshow(gray, cmap="gray")
plt.title("Gray")
plt.axis("off")

# HSV
plt.subplot(1, 4, 4)
plt.imshow(hsv_rgb)
plt.title("HSV")
plt.axis("off")

plt.show()
```

```python
#تنعيم 
blur = cv2.blur(image, (5, 5))

gaussian = cv2.GaussianBlur(image, (5, 5), 0)

median = cv2.medianBlur(image, 5)
```

```python
import matplotlib.pyplot as plt
import cv2

plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(cv2.cvtColor(blur, cv2.COLOR_BGR2RGB))
plt.title("Average Blur")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Blur")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(cv2.cvtColor(median, cv2.COLOR_BGR2RGB))
plt.title("Median Blur")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

image = cv2.imread(r"C:\Users\PC-LAB1\Downloads\WhatsApp Image 2026-09-22 at 10.59.56 AM.jpeg")

median = cv2.medianBlur(image, 5)

# الصورة قبل
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Before")
plt.axis("off")
plt.show()

# الصورة بعد
plt.imshow(cv2.cvtColor(median, cv2.COLOR_BGR2RGB))
plt.title("After")
plt.axis("off")
plt.show()

# Histogram قبل
plt.hist(image.ravel(), 256)
plt.title("Histogram Before")
plt.show()

# Histogram بعد
plt.hist(median.ravel(), 256)
plt.title("Histogram After")
plt.show()
```
