```python
import os
import cv2
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.models import load_model

from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout,
    Rescaling
)


# ==========================================
# 1 - Settings
# ==========================================

dataset_path = r"C:\Users\PC-LAB1\Desktop\New folder\day 5\dogs-vs-cats\train\train"

IMG_SIZE = 128

TOTAL_PER_CLASS = 1000

random.seed(42)

np.random.seed(42)

tf.random.set_seed(42)


# ==========================================
# 2 - Get Image Files
# ==========================================

all_files = os.listdir(dataset_path)

cat_files = [
    f for f in all_files
    if f.lower().startswith("cat.")
    and f.lower().endswith(
        (".jpg", ".jpeg", ".png")
    )
]

dog_files = [
    f for f in all_files
    if f.lower().startswith("dog.")
    and f.lower().endswith(
        (".jpg", ".jpeg", ".png")
    )
]


print("Total Cat images:", len(cat_files))

print("Total Dog images:", len(dog_files))


# ==========================================
# 3 - Select 1000 Cats + 1000 Dogs
# ==========================================

random.shuffle(cat_files)

random.shuffle(dog_files)

cat_files = cat_files[:TOTAL_PER_CLASS]

dog_files = dog_files[:TOTAL_PER_CLASS]


print()
print("Selected Cats:", len(cat_files))

print("Selected Dogs:", len(dog_files))


# ==========================================
# 4 - Load Images
# ==========================================

X = []

y = []


# ---------- Cats = 0 ----------

for image_name in cat_files:

    image_path = os.path.join(
        dataset_path,
        image_name
    )

    image = cv2.imread(image_path)

    if image is None:
        continue

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    image = cv2.resize(
        image,
        (IMG_SIZE, IMG_SIZE)
    )

    X.append(image)

    y.append(0)


# ---------- Dogs = 1 ----------

for image_name in dog_files:

    image_path = os.path.join(
        dataset_path,
        image_name
    )

    image = cv2.imread(image_path)

    if image is None:
        continue

    image = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2RGB
    )

    image = cv2.resize(
        image,
        (IMG_SIZE, IMG_SIZE)
    )

    X.append(image)

    y.append(1)


# ==========================================
# 5 - Convert to NumPy
# ==========================================

X = np.array(X)

y = np.array(y)


print()
print("X shape:", X.shape)

print("y shape:", y.shape)


# ==========================================
# 6 - Train / Test Split
# ==========================================

X_train, X_test, y_train, y_test = train_test_split(

    X,
    y,

    test_size=0.20,

    random_state=42,

    stratify=y
)


print()
print("Training images:", len(X_train))

print("Testing images:", len(X_test))


# ==========================================
# 7 - Build CNN
# ==========================================

model = Sequential([

    Rescaling(
        1.0 / 255
    ),

    Conv2D(
        32,
        (3, 3),
        activation="relu",
        input_shape=(128, 128, 3)
    ),

    MaxPooling2D(
        (2, 2)
    ),

    Conv2D(
        64,
        (3, 3),
        activation="relu"
    ),

    MaxPooling2D(
        (2, 2)
    ),

    Conv2D(
        128,
        (3, 3),
        activation="relu"
    ),

    MaxPooling2D(
        (2, 2)
    ),

    Flatten(),

    Dense(
        128,
        activation="relu"
    ),

    Dropout(
        0.5
    ),

    Dense(
        1,
        activation="sigmoid"
    )
])


# ==========================================
# 8 - Model Summary
# ==========================================

model.summary()


# ==========================================
# 9 - Compile
# ==========================================

model.compile(

    optimizer="adam",

    loss="binary_crossentropy",

    metrics=["accuracy"]
)


# ==========================================
# 10 - Training
# ==========================================

history = model.fit(

    X_train,

    y_train,

    epochs=10,

    batch_size=32,

    validation_split=0.20,

    shuffle=True
)


# ==========================================
# 11 - Accuracy Curve
# ==========================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title(
    "CNN Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.legend()

plt.grid()

plt.show()


# ==========================================
# 12 - Loss Curve
# ==========================================

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.title(
    "CNN Loss"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.legend()

plt.grid()

plt.show()


# ==========================================
# 13 - Test Evaluation
# ==========================================

test_loss, test_accuracy = model.evaluate(

    X_test,

    y_test,

    verbose=1
)


print()
print("================================")
print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)
print(
    "Test Accuracy %:",
    test_accuracy * 100
)
print("================================")


# ==========================================
# 14 - Predictions
# ==========================================

predictions = model.predict(
    X_test
)


y_pred = (
    predictions >= 0.5
).astype(int)


y_pred = y_pred.flatten()


# ==========================================
# 15 - Accuracy
# ==========================================

accuracy = accuracy_score(

    y_test,

    y_pred
)


print()
print(
    "Accuracy:",
    accuracy
)

print(
    "Accuracy %:",
    accuracy * 100
)


# ==========================================
# 16 - Classification Report
# ==========================================

print()
print("========================================")

print("Classification Report")

print("========================================")


print(
    classification_report(

        y_test,

        y_pred,

        target_names=[
            "Cat",
            "Dog"
        ]
    )
)


# ==========================================
# 17 - Confusion Matrix
# ==========================================

cm = confusion_matrix(

    y_test,

    y_pred
)


print()
print("Confusion Matrix:")

print(cm)


plt.figure(
    figsize=(7, 6)
)


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=[
        "Cat",
        "Dog"
    ],

    yticklabels=[
        "Cat",
        "Dog"
    ]
)


plt.title(
    "CNN Confusion Matrix"
)

plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)

plt.show()


# ==========================================
# 18 - Display 10 Test Images
# ==========================================

random_indices = random.sample(

    range(len(X_test)),

    10
)


plt.figure(
    figsize=(15, 8)
)


for i, index in enumerate(
    random_indices
):

    plt.subplot(
        2,
        5,
        i + 1
    )


    plt.imshow(
        X_test[index]
    )


    actual = (
        "Cat"
        if y_test[index] == 0
        else "Dog"
    )


    predicted = (
        "Cat"
        if y_pred[index] == 0
        else "Dog"
    )


    plt.title(
        f"Actual: {actual}\n"
        f"Predicted: {predicted}"
    )


    plt.axis(
        "off"
    )


plt.tight_layout()

plt.show()


# ==========================================
# 19 - Save CSV
# ==========================================

test_results = []


for i in range(
    len(X_test)
):

    actual = (
        "Cat"
        if y_test[i] == 0
        else "Dog"
    )


    predicted = (
        "Cat"
        if y_pred[i] == 0
        else "Dog"
    )


    probability = float(
        predictions[i][0]
    )


    test_results.append([

        i,

        actual,

        predicted,

        probability

    ])


results_df = pd.DataFrame(

    test_results,

    columns=[

        "Image_ID",

        "Actual",

        "Predicted",

        "Dog_Probability"

    ]
)


results_df.to_csv(

    "dogs_cats_cnn_test_results.csv",

    index=False
)


print()
print(
    "CSV saved successfully!"
)


# ==========================================
# 20 - Save CNN Model
# ==========================================

model.save(

    "dogs_cats_cnn_model.keras"
)


print(
    "CNN model saved successfully!"
)
```

```python
"""
Cats vs Dogs - CNN Classifier (Professional Version)
====================================================

Usage examples:
    python cats_vs_dogs_cnn.py
    python cats_vs_dogs_cnn.py --data "C:\\path\\to\\train" --epochs 30 --show
    python cats_vs_dogs_cnn.py --predict "C:\\path\\to\\image.jpg" --model outputs\\run_xxx\\best_model.keras
"""

from __future__ import annotations

import argparse
import json
import logging
import random
from concurrent.futures import ThreadPoolExecutor
from dataclasses import dataclass, asdict
from datetime import datetime
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import tensorflow as tf
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    roc_curve,
)
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.callbacks import (
    CSVLogger,
    EarlyStopping,
    ModelCheckpoint,
    ReduceLROnPlateau,
)

logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(levelname)s | %(message)s")
log = logging.getLogger("cats_dogs")

CLASS_NAMES = ["Cat", "Dog"]  # label 0 = Cat, label 1 = Dog
VALID_EXT = (".jpg", ".jpeg", ".png")


# ============================================================
# 1 - CONFIGURATION
# ============================================================
@dataclass
class Config:
    data_dir: str = r"C:\Users\PC-LAB1\Desktop\New folder\day 5\dogs-vs-cats\train\train"
    output_dir: str = "outputs"
    img_size: int = 128
    per_class: int = 1000
    test_size: float = 0.20
    val_size: float = 0.20          # fraction of the remaining train data
    batch_size: int = 32
    epochs: int = 40
    learning_rate: float = 1e-3
    dropout: float = 0.5
    l2: float = 1e-4
    seed: int = 42
    workers: int = 8
    show: bool = False              # show plots on screen (they are always saved)


def parse_args() -> tuple[Config, str | None, str | None]:
    d = Config()
    p = argparse.ArgumentParser(description="Cats vs Dogs CNN")
    p.add_argument("--data", dest="data_dir", default=d.data_dir)
    p.add_argument("--output", dest="output_dir", default=d.output_dir)
    p.add_argument("--img-size", type=int, default=d.img_size)
    p.add_argument("--per-class", type=int, default=d.per_class)
    p.add_argument("--batch-size", type=int, default=d.batch_size)
    p.add_argument("--epochs", type=int, default=d.epochs)
    p.add_argument("--lr", dest="learning_rate", type=float, default=d.learning_rate)
    p.add_argument("--seed", type=int, default=d.seed)
    p.add_argument("--show", action="store_true", help="display plots in windows")
    p.add_argument("--predict", default=None, help="path of a new image to classify")
    p.add_argument("--model", default=None, help="path of a saved .keras model (for --predict)")
    # parse_known_args: ignores extra args injected by Jupyter/IPython (e.g. --f=kernel.json)
    a, _ = p.parse_known_args()
    cfg = Config(
        data_dir=a.data_dir, output_dir=a.output_dir, img_size=a.img_size,
        per_class=a.per_class, batch_size=a.batch_size, epochs=a.epochs,
        learning_rate=a.learning_rate, seed=a.seed, show=a.show,
    )
    return cfg, a.predict, a.model


def set_seed(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)
    tf.keras.utils.set_random_seed(seed)


# ============================================================
# 2 - DATA LOADING
# ============================================================
def load_image(path: Path, size: int) -> np.ndarray | None:
    """Read an image as RGB uint8 (size x size). Returns None if unreadable."""
    img = cv2.imread(str(path))
    if img is None:
        return None
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    return cv2.resize(img, (size, size), interpolation=cv2.INTER_AREA)


def select_files(data_dir: Path, prefix: str, n: int, rng: random.Random) -> list[Path]:
    files = sorted(
        f for f in data_dir.iterdir()
        if f.name.lower().startswith(prefix) and f.suffix.lower() in VALID_EXT
    )
    if len(files) < n:
        raise ValueError(f"Not enough '{prefix}' images: found {len(files)}, need {n}")
    return rng.sample(files, n)


def load_dataset(cfg: Config) -> tuple[np.ndarray, np.ndarray, np.ndarray]:
    data_dir = Path(cfg.data_dir)
    if not data_dir.is_dir():
        raise FileNotFoundError(f"Dataset path not found: {data_dir}")

    rng = random.Random(cfg.seed)
    cats = select_files(data_dir, "cat.", cfg.per_class, rng)
    dogs = select_files(data_dir, "dog.", cfg.per_class, rng)
    paths = cats + dogs
    labels = [0] * len(cats) + [1] * len(dogs)

    log.info("Loading %d images with %d threads...", len(paths), cfg.workers)
    with ThreadPoolExecutor(max_workers=cfg.workers) as pool:
        images = list(pool.map(lambda p: load_image(p, cfg.img_size), paths))

    keep = [i for i, im in enumerate(images) if im is not None]
    skipped = len(paths) - len(keep)
    if skipped:
        log.warning("Skipped %d unreadable images", skipped)

    X = np.stack([images[i] for i in keep]).astype(np.uint8)   # uint8 = 4x less RAM
    y = np.array([labels[i] for i in keep], dtype=np.int32)
    names = np.array([paths[i].name for i in keep])
    log.info("Loaded X=%s | cats=%d | dogs=%d", X.shape, (y == 0).sum(), (y == 1).sum())
    return X, y, names


def split_data(X, y, names, cfg: Config):
    """Stratified train / validation / test split."""
    idx = np.arange(len(y))
    idx_trainval, idx_test = train_test_split(
        idx, test_size=cfg.test_size, random_state=cfg.seed, stratify=y
    )
    idx_train, idx_val = train_test_split(
        idx_trainval, test_size=cfg.val_size, random_state=cfg.seed, stratify=y[idx_trainval]
    )
    parts = {"train": idx_train, "val": idx_val, "test": idx_test}
    for k, v in parts.items():
        log.info("%-5s: %4d images | cats=%d dogs=%d", k, len(v), (y[v] == 0).sum(), (y[v] == 1).sum())
    return {k: (X[v], y[v], names[v]) for k, v in parts.items()}


def make_dataset(X, y, cfg: Config, training: bool) -> tf.data.Dataset:
    ds = tf.data.Dataset.from_tensor_slices((X, y))
    if training:
        ds = ds.shuffle(len(y), seed=cfg.seed, reshuffle_each_iteration=True)
    ds = ds.batch(cfg.batch_size)
    ds = ds.map(lambda a, b: (tf.cast(a, tf.float32), tf.cast(b, tf.float32)),
                num_parallel_calls=tf.data.AUTOTUNE)
    return ds.prefetch(tf.data.AUTOTUNE)


# ============================================================
# 3 - MODEL
# ============================================================
def build_model(cfg: Config) -> tf.keras.Model:
    reg = regularizers.l2(cfg.l2)

    augmentation = tf.keras.Sequential(
        [layers.RandomFlip("horizontal"), layers.RandomRotation(0.10), layers.RandomZoom(0.10)],
        name="augmentation",
    )

    inputs = layers.Input(shape=(cfg.img_size, cfg.img_size, 3))
    x = layers.Rescaling(1.0 / 255)(inputs)
    x = augmentation(x)                       # active only during training

    for filters in (32, 64, 128, 256):
        x = layers.Conv2D(filters, 3, padding="same", use_bias=False, kernel_regularizer=reg)(x)
        x = layers.BatchNormalization()(x)
        x = layers.ReLU()(x)
        x = layers.MaxPooling2D()(x)

    x = layers.GlobalAveragePooling2D()(x)    # far fewer parameters than Flatten
    x = layers.Dense(128, activation="relu", kernel_regularizer=reg)(x)
    x = layers.Dropout(cfg.dropout)(x)
    outputs = layers.Dense(1, activation="sigmoid")(x)   # Cat=0, Dog=1

    model = models.Model(inputs, outputs, name="cats_vs_dogs_cnn")
    model.compile(
        optimizer=tf.keras.optimizers.Adam(cfg.learning_rate),
        loss="binary_crossentropy",
        metrics=[
            "accuracy",
            tf.keras.metrics.AUC(name="auc"),
            tf.keras.metrics.Precision(name="precision"),
            tf.keras.metrics.Recall(name="recall"),
        ],
    )
    return model


# ============================================================
# 4 - PLOTS (always saved, optionally shown)
# ============================================================
def finish_fig(fig, path: Path, show: bool) -> None:
    fig.tight_layout()
    fig.savefig(path, dpi=150)
    if show:
        plt.show()
    plt.close(fig)


def plot_history(history: dict, out: Path, show: bool) -> None:
    fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
    for ax, metric, title in zip(axes, ("accuracy", "loss"), ("Accuracy", "Loss")):
        ax.plot(history[metric], label=f"Train {title}")
        ax.plot(history[f"val_{metric}"], label=f"Val {title}")
        ax.set(title=f"Training vs Validation {title}", xlabel="Epoch", ylabel=title)
        ax.legend()
        ax.grid(alpha=0.3)
    finish_fig(fig, out / "training_curves.png", show)


def plot_confusion(cm, out: Path, show: bool) -> None:
    fig, ax = plt.subplots(figsize=(5.5, 4.5))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=CLASS_NAMES,
                yticklabels=CLASS_NAMES, ax=ax)
    ax.set(xlabel="Predicted", ylabel="Actual", title="Confusion Matrix")
    finish_fig(fig, out / "confusion_matrix.png", show)


def plot_roc(y_true, probs, out: Path, show: bool) -> None:
    fpr, tpr, _ = roc_curve(y_true, probs)
    auc = roc_auc_score(y_true, probs)
    fig, ax = plt.subplots(figsize=(5.5, 5))
    ax.plot(fpr, tpr, label=f"AUC = {auc:.4f}")
    ax.plot([0, 1], [0, 1], "--", color="gray")
    ax.set(xlabel="False Positive Rate", ylabel="True Positive Rate", title="ROC Curve")
    ax.legend()
    ax.grid(alpha=0.3)
    finish_fig(fig, out / "roc_curve.png", show)


def plot_samples(X, y_true, y_pred, probs, indices, title, path: Path, show: bool) -> None:
    indices = list(indices)[:10]
    if not indices:
        return
    fig, axes = plt.subplots(2, 5, figsize=(15, 7))
    for ax in axes.ravel():
        ax.axis("off")
    for ax, i in zip(axes.ravel(), indices):
        conf = probs[i] if y_pred[i] == 1 else 1 - probs[i]
        ax.imshow(X[i])
        ax.set_title(f"Actual: {CLASS_NAMES[y_true[i]]}\nPred: {CLASS_NAMES[y_pred[i]]} "
                     f"({conf * 100:.1f}%)", fontsize=9,
                     color="green" if y_true[i] == y_pred[i] else "red")
    fig.suptitle(title)
    finish_fig(fig, path, show)


# ============================================================
# 5 - EVALUATION
# ============================================================
def evaluate(model, test_data, cfg: Config, out: Path) -> dict:
    X_test, y_test, names_test = test_data
    test_ds = make_dataset(X_test, y_test, cfg, training=False)

    metrics = model.evaluate(test_ds, return_dict=True, verbose=0)
    probs = model.predict(test_ds, verbose=0).ravel()
    y_pred = (probs >= 0.5).astype(int)

    cm = confusion_matrix(y_test, y_pred)
    report = classification_report(y_test, y_pred, target_names=CLASS_NAMES, digits=4)
    log.info("Test metrics: %s", {k: round(float(v), 4) for k, v in metrics.items()})
    print("\n" + report)
    print("Confusion matrix:\n", cm)

    plot_confusion(cm, out, cfg.show)
    plot_roc(y_test, probs, out, cfg.show)
    plot_samples(X_test, y_test, y_pred, probs, range(len(y_test)),
                 "Random test predictions", out / "test_samples.png", cfg.show)
    wrong = np.where(y_pred != y_test)[0]
    plot_samples(X_test, y_test, y_pred, probs, wrong,
                 f"Misclassified images ({len(wrong)} total)", out / "misclassified.png", cfg.show)

    pd.DataFrame({
        "file": names_test,
        "actual": [CLASS_NAMES[i] for i in y_test],
        "predicted": [CLASS_NAMES[i] for i in y_pred],
        "probability_dog": probs,
        "confidence": np.where(y_pred == 1, probs, 1 - probs),
        "correct": y_pred == y_test,
    }).to_csv(out / "test_predictions.csv", index=False)

    (out / "test_report.txt").write_text(report, encoding="utf-8")
    return {k: float(v) for k, v in metrics.items()}


# ============================================================
# 6 - INFERENCE ON A NEW IMAGE
# ============================================================
def predict_image(model: tf.keras.Model, image_path: str, img_size: int, show: bool = True) -> None:
    path = Path(image_path)
    if not path.is_file():
        log.error("Image not found: %s", path)
        return
    img = load_image(path, img_size)
    if img is None:
        log.error("Could not read image: %s", path)
        return

    prob = float(model.predict(img[np.newaxis].astype(np.float32), verbose=0)[0][0])
    label = CLASS_NAMES[int(prob >= 0.5)]
    conf = (prob if prob >= 0.5 else 1 - prob) * 100
    log.info("Prediction: %s (%.2f%% confidence)", label, conf)

    if show:
        plt.imshow(img)
        plt.title(f"{label} - {conf:.2f}%")
        plt.axis("off")
        plt.show()


# ============================================================
# 7 - MAIN
# ============================================================
def main() -> None:
    cfg, predict_path, model_path = parse_args()
    set_seed(cfg.seed)

    # ---- inference-only mode ----
    if predict_path:
        if not model_path:
            raise SystemExit("--predict requires --model <path to .keras file>")
        predict_image(models.load_model(model_path), predict_path, cfg.img_size)
        return

    # ---- training mode ----
    out = Path(cfg.output_dir) / f"run_{datetime.now():%Y%m%d_%H%M%S}"
    out.mkdir(parents=True, exist_ok=True)
    (out / "config.json").write_text(json.dumps(asdict(cfg), indent=2), encoding="utf-8")
    log.info("Outputs will be saved in: %s", out.resolve())

    X, y, names = load_dataset(cfg)
    splits = split_data(X, y, names, cfg)

    plot_samples(X, y, y, np.where(y == 1, 1.0, 0.0), np.random.permutation(len(y))[:10],
                 "Random training samples", out / "dataset_samples.png", cfg.show)

    train_ds = make_dataset(*splits["train"][:2], cfg, training=True)
    val_ds = make_dataset(*splits["val"][:2], cfg, training=False)

    weights = compute_class_weight("balanced", classes=np.array([0, 1]), y=splits["train"][1])
    class_weight = {0: float(weights[0]), 1: float(weights[1])}
    log.info("Class weights: %s", class_weight)

    model = build_model(cfg)
    model.summary()

    best_path = out / "best_model.keras"
    callbacks = [
        EarlyStopping(monitor="val_loss", patience=8, restore_best_weights=True, verbose=1),
        ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, min_lr=1e-6, verbose=1),
        ModelCheckpoint(str(best_path), monitor="val_loss", save_best_only=True),
        CSVLogger(str(out / "training_log.csv")),
    ]

    history = model.fit(
        train_ds, validation_data=val_ds, epochs=cfg.epochs,
        class_weight=class_weight, callbacks=callbacks, verbose=1,
    )
    plot_history(history.history, out, cfg.show)

    results = evaluate(model, splits["test"], cfg, out)
    (out / "test_metrics.json").write_text(json.dumps(results, indent=2), encoding="utf-8")

    # verify the saved model reloads and gives the same predictions
    reloaded = models.load_model(best_path)
    sample = splits["test"][0][:8].astype(np.float32)
    assert np.allclose(model.predict(sample, verbose=0), reloaded.predict(sample, verbose=0), atol=1e-5), \
        "Reloaded model output differs from the trained model!"
    log.info("Saved model verified: %s", best_path)

    log.info("Test accuracy: %.2f%% | AUC: %.4f", results["accuracy"] * 100, results["auc"])
    log.info("DONE. All files are in: %s", out.resolve())


if __name__ == "__main__":
    main()
```

```python
"""
Cats vs Dogs - CNN Classifier (Professional Version)
====================================================

Usage examples:
    python cats_vs_dogs_cnn.py
    python cats_vs_dogs_cnn.py --data "C:\\path\\to\\train" --epochs 30 --show
    python cats_vs_dogs_cnn.py --predict "C:\\path\\to\\image.jpg" --model outputs\\run_xxx\\best_model.keras
"""

from __future__ import annotations

import argparse
import json
import logging
import random
from concurrent.futures import ThreadPoolExecutor
from dataclasses import dataclass, asdict
from datetime import datetime
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import tensorflow as tf
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    roc_curve,
)
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.callbacks import (
    CSVLogger,
    EarlyStopping,
    ModelCheckpoint,
    ReduceLROnPlateau,
)

logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(levelname)s | %(message)s")
log = logging.getLogger("cats_dogs")

CLASS_NAMES = ["Cat", "Dog"]  # label 0 = Cat, label 1 = Dog
VALID_EXT = (".jpg", ".jpeg", ".png")


# ============================================================
# 1 - CONFIGURATION
# ============================================================
@dataclass
class Config:
    data_dir: str = r"C:\Users\PC-LAB1\Desktop\New folder\day 5\dogs-vs-cats\train\train"
    output_dir: str = "outputs"
    img_size: int = 128
    per_class: int = 0              # 0 = use ALL images (balanced to the smaller class)
    test_size: float = 0.20
    val_size: float = 0.20          # fraction of the remaining train data
    batch_size: int = 64
    epochs: int = 10
    learning_rate: float = 1e-3
    dropout: float = 0.5
    l2: float = 1e-4
    seed: int = 42
    workers: int = 8
    show: bool = True               # show plots with plt.show() (they are always saved too)


def parse_args() -> tuple[Config, str | None, str | None]:
    d = Config()
    p = argparse.ArgumentParser(description="Cats vs Dogs CNN")
    p.add_argument("--data", dest="data_dir", default=d.data_dir)
    p.add_argument("--output", dest="output_dir", default=d.output_dir)
    p.add_argument("--img-size", type=int, default=d.img_size)
    p.add_argument("--per-class", type=int, default=d.per_class)
    p.add_argument("--batch-size", type=int, default=d.batch_size)
    p.add_argument("--epochs", type=int, default=d.epochs)
    p.add_argument("--lr", dest="learning_rate", type=float, default=d.learning_rate)
    p.add_argument("--seed", type=int, default=d.seed)
    p.add_argument("--no-show", action="store_true", help="do not display plots (only save them)")
    p.add_argument("--predict", default=None, help="path of a new image to classify")
    p.add_argument("--model", default=None, help="path of a saved .keras model (for --predict)")
    # parse_known_args: ignores extra args injected by Jupyter/IPython (e.g. --f=kernel.json)
    a, _ = p.parse_known_args()
    cfg = Config(
        data_dir=a.data_dir, output_dir=a.output_dir, img_size=a.img_size,
        per_class=a.per_class, batch_size=a.batch_size, epochs=a.epochs,
        learning_rate=a.learning_rate, seed=a.seed, show=not a.no_show,
    )
    return cfg, a.predict, a.model


def set_seed(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)
    tf.keras.utils.set_random_seed(seed)


# ============================================================
# 2 - DATA LOADING
# ============================================================
def load_image(path: Path, size: int) -> np.ndarray | None:
    """Read an image as RGB uint8 (size x size). Returns None if unreadable."""
    img = cv2.imread(str(path))
    if img is None:
        return None
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    return cv2.resize(img, (size, size), interpolation=cv2.INTER_AREA)


def list_files(data_dir: Path, prefix: str) -> list[Path]:
    return sorted(
        f for f in data_dir.iterdir()
        if f.name.lower().startswith(prefix) and f.suffix.lower() in VALID_EXT
    )


def select_files(files: list[Path], n: int, rng: random.Random) -> list[Path]:
    if len(files) < n:
        raise ValueError(f"Not enough images: found {len(files)}, need {n}")
    return rng.sample(files, n)


def load_dataset(cfg: Config) -> tuple[np.ndarray, np.ndarray, np.ndarray]:
    data_dir = Path(cfg.data_dir)
    if not data_dir.is_dir():
        raise FileNotFoundError(f"Dataset path not found: {data_dir}")

    rng = random.Random(cfg.seed)
    all_cats = list_files(data_dir, "cat.")
    all_dogs = list_files(data_dir, "dog.")
    log.info("Available: %d cats, %d dogs", len(all_cats), len(all_dogs))

    # per_class = 0  ->  use every image (balanced to the smaller class)
    n = cfg.per_class if cfg.per_class > 0 else min(len(all_cats), len(all_dogs))
    cats = select_files(all_cats, n, rng)
    dogs = select_files(all_dogs, n, rng)
    paths = cats + dogs
    labels = [0] * len(cats) + [1] * len(dogs)

    log.info("Loading %d images with %d threads...", len(paths), cfg.workers)
    with ThreadPoolExecutor(max_workers=cfg.workers) as pool:
        images = list(pool.map(lambda p: load_image(p, cfg.img_size), paths))

    keep = [i for i, im in enumerate(images) if im is not None]
    skipped = len(paths) - len(keep)
    if skipped:
        log.warning("Skipped %d unreadable images", skipped)

    X = np.stack([images[i] for i in keep]).astype(np.uint8)   # uint8 = 4x less RAM
    y = np.array([labels[i] for i in keep], dtype=np.int32)
    names = np.array([paths[i].name for i in keep])
    log.info("Loaded X=%s | cats=%d | dogs=%d", X.shape, (y == 0).sum(), (y == 1).sum())
    return X, y, names


def split_data(X, y, names, cfg: Config):
    """Stratified train / validation / test split."""
    idx = np.arange(len(y))
    idx_trainval, idx_test = train_test_split(
        idx, test_size=cfg.test_size, random_state=cfg.seed, stratify=y
    )
    idx_train, idx_val = train_test_split(
        idx_trainval, test_size=cfg.val_size, random_state=cfg.seed, stratify=y[idx_trainval]
    )
    parts = {"train": idx_train, "val": idx_val, "test": idx_test}
    for k, v in parts.items():
        log.info("%-5s: %4d images | cats=%d dogs=%d", k, len(v), (y[v] == 0).sum(), (y[v] == 1).sum())
    return {k: (X[v], y[v], names[v]) for k, v in parts.items()}


def make_dataset(X, y, cfg: Config, training: bool) -> tf.data.Dataset:
    ds = tf.data.Dataset.from_tensor_slices((X, y))
    if training:
        ds = ds.shuffle(len(y), seed=cfg.seed, reshuffle_each_iteration=True)
    ds = ds.batch(cfg.batch_size)
    ds = ds.map(lambda a, b: (tf.cast(a, tf.float32), tf.cast(b, tf.float32)),
                num_parallel_calls=tf.data.AUTOTUNE)
    return ds.prefetch(tf.data.AUTOTUNE)


# ============================================================
# 3 - MODEL
# ============================================================
def build_model(cfg: Config) -> tf.keras.Model:
    reg = regularizers.l2(cfg.l2)

    augmentation = tf.keras.Sequential(
        [layers.RandomFlip("horizontal"), layers.RandomRotation(0.10), layers.RandomZoom(0.10)],
        name="augmentation",
    )

    inputs = layers.Input(shape=(cfg.img_size, cfg.img_size, 3))
    x = layers.Rescaling(1.0 / 255)(inputs)
    x = augmentation(x)                       # active only during training

    for filters in (32, 64, 128, 256):
        x = layers.Conv2D(filters, 3, padding="same", use_bias=False, kernel_regularizer=reg)(x)
        x = layers.BatchNormalization()(x)
        x = layers.ReLU()(x)
        x = layers.MaxPooling2D()(x)

    x = layers.GlobalAveragePooling2D()(x)    # far fewer parameters than Flatten
    x = layers.Dense(128, activation="relu", kernel_regularizer=reg)(x)
    x = layers.Dropout(cfg.dropout)(x)
    outputs = layers.Dense(1, activation="sigmoid")(x)   # Cat=0, Dog=1

    model = models.Model(inputs, outputs, name="cats_vs_dogs_cnn")
    model.compile(
        optimizer=tf.keras.optimizers.Adam(cfg.learning_rate),
        loss="binary_crossentropy",
        metrics=[
            "accuracy",
            tf.keras.metrics.AUC(name="auc"),
            tf.keras.metrics.Precision(name="precision"),
            tf.keras.metrics.Recall(name="recall"),
        ],
    )
    return model


# ============================================================
# 4 - PLOTS (always saved, optionally shown)
# ============================================================
def finish_fig(fig, path: Path, show: bool) -> None:
    fig.tight_layout()
    fig.savefig(path, dpi=150)
    if show:
        plt.show()
    plt.close(fig)


def plot_history(history: dict, out: Path, show: bool) -> None:
    fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
    for ax, metric, title in zip(axes, ("accuracy", "loss"), ("Accuracy", "Loss")):
        ax.plot(history[metric], label=f"Train {title}")
        ax.plot(history[f"val_{metric}"], label=f"Val {title}")
        ax.set(title=f"Training vs Validation {title}", xlabel="Epoch", ylabel=title)
        ax.legend()
        ax.grid(alpha=0.3)
    finish_fig(fig, out / "training_curves.png", show)


def plot_confusion(cm, out: Path, show: bool) -> None:
    fig, ax = plt.subplots(figsize=(5.5, 4.5))
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=CLASS_NAMES,
                yticklabels=CLASS_NAMES, ax=ax)
    ax.set(xlabel="Predicted", ylabel="Actual", title="Confusion Matrix")
    finish_fig(fig, out / "confusion_matrix.png", show)


def plot_roc(y_true, probs, out: Path, show: bool) -> None:
    fpr, tpr, _ = roc_curve(y_true, probs)
    auc = roc_auc_score(y_true, probs)
    fig, ax = plt.subplots(figsize=(5.5, 5))
    ax.plot(fpr, tpr, label=f"AUC = {auc:.4f}")
    ax.plot([0, 1], [0, 1], "--", color="gray")
    ax.set(xlabel="False Positive Rate", ylabel="True Positive Rate", title="ROC Curve")
    ax.legend()
    ax.grid(alpha=0.3)
    finish_fig(fig, out / "roc_curve.png", show)


def plot_samples(X, y_true, y_pred, probs, indices, title, path: Path, show: bool) -> None:
    indices = list(indices)[:10]
    if not indices:
        return
    fig, axes = plt.subplots(2, 5, figsize=(15, 7))
    for ax in axes.ravel():
        ax.axis("off")
    for ax, i in zip(axes.ravel(), indices):
        conf = probs[i] if y_pred[i] == 1 else 1 - probs[i]
        ax.imshow(X[i])
        ax.set_title(f"Actual: {CLASS_NAMES[y_true[i]]}\nPred: {CLASS_NAMES[y_pred[i]]} "
                     f"({conf * 100:.1f}%)", fontsize=9,
                     color="green" if y_true[i] == y_pred[i] else "red")
    fig.suptitle(title)
    finish_fig(fig, path, show)


# ============================================================
# 5 - EVALUATION
# ============================================================
def evaluate(model, test_data, cfg: Config, out: Path) -> dict:
    X_test, y_test, names_test = test_data
    test_ds = make_dataset(X_test, y_test, cfg, training=False)

    metrics = model.evaluate(test_ds, return_dict=True, verbose=0)
    probs = model.predict(test_ds, verbose=0).ravel()
    y_pred = (probs >= 0.5).astype(int)

    cm = confusion_matrix(y_test, y_pred)
    report = classification_report(y_test, y_pred, target_names=CLASS_NAMES, digits=4)
    log.info("Test metrics: %s", {k: round(float(v), 4) for k, v in metrics.items()})
    print("\n" + report)
    print("Confusion matrix:\n", cm)

    plot_confusion(cm, out, cfg.show)
    plot_roc(y_test, probs, out, cfg.show)
    plot_samples(X_test, y_test, y_pred, probs, range(len(y_test)),
                 "Random test predictions", out / "test_samples.png", cfg.show)
    wrong = np.where(y_pred != y_test)[0]
    plot_samples(X_test, y_test, y_pred, probs, wrong,
                 f"Misclassified images ({len(wrong)} total)", out / "misclassified.png", cfg.show)

    pd.DataFrame({
        "file": names_test,
        "actual": [CLASS_NAMES[i] for i in y_test],
        "predicted": [CLASS_NAMES[i] for i in y_pred],
        "probability_dog": probs,
        "confidence": np.where(y_pred == 1, probs, 1 - probs),
        "correct": y_pred == y_test,
    }).to_csv(out / "test_predictions.csv", index=False)

    (out / "test_report.txt").write_text(report, encoding="utf-8")
    return {k: float(v) for k, v in metrics.items()}


# ============================================================
# 6 - INFERENCE ON A NEW IMAGE
# ============================================================
def predict_image(model: tf.keras.Model, image_path: str, img_size: int, show: bool = True) -> None:
    path = Path(image_path)
    if not path.is_file():
        log.error("Image not found: %s", path)
        return
    img = load_image(path, img_size)
    if img is None:
        log.error("Could not read image: %s", path)
        return

    prob = float(model.predict(img[np.newaxis].astype(np.float32), verbose=0)[0][0])
    label = CLASS_NAMES[int(prob >= 0.5)]
    conf = (prob if prob >= 0.5 else 1 - prob) * 100
    log.info("Prediction: %s (%.2f%% confidence)", label, conf)

    if show:
        plt.imshow(img)
        plt.title(f"{label} - {conf:.2f}%")
        plt.axis("off")
        plt.show()


# ============================================================
# 7 - MAIN
# ============================================================
def main() -> None:
    cfg, predict_path, model_path = parse_args()
    set_seed(cfg.seed)

    # ---- inference-only mode ----
    if predict_path:
        if not model_path:
            raise SystemExit("--predict requires --model <path to .keras file>")
        predict_image(models.load_model(model_path), predict_path, cfg.img_size)
        return

    # ---- training mode ----
    out = Path(cfg.output_dir) / f"run_{datetime.now():%Y%m%d_%H%M%S}"
    out.mkdir(parents=True, exist_ok=True)
    (out / "config.json").write_text(json.dumps(asdict(cfg), indent=2), encoding="utf-8")
    log.info("Outputs will be saved in: %s", out.resolve())

    X, y, names = load_dataset(cfg)
    splits = split_data(X, y, names, cfg)

    plot_samples(X, y, y, np.where(y == 1, 1.0, 0.0), np.random.permutation(len(y))[:10],
                 "Random training samples", out / "dataset_samples.png", cfg.show)

    train_ds = make_dataset(*splits["train"][:2], cfg, training=True)
    val_ds = make_dataset(*splits["val"][:2], cfg, training=False)

    weights = compute_class_weight("balanced", classes=np.array([0, 1]), y=splits["train"][1])
    class_weight = {0: float(weights[0]), 1: float(weights[1])}
    log.info("Class weights: %s", class_weight)

    model = build_model(cfg)
    model.summary()

    best_path = out / "best_model.keras"
    callbacks = [
        EarlyStopping(monitor="val_loss", patience=8, restore_best_weights=True, verbose=1),
        ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, min_lr=1e-6, verbose=1),
        ModelCheckpoint(str(best_path), monitor="val_loss", save_best_only=True),
        CSVLogger(str(out / "training_log.csv")),
    ]

    history = model.fit(
        train_ds, validation_data=val_ds, epochs=cfg.epochs,
        class_weight=class_weight, callbacks=callbacks, verbose=1,
    )
    plot_history(history.history, out, cfg.show)

    results = evaluate(model, splits["test"], cfg, out)
    (out / "test_metrics.json").write_text(json.dumps(results, indent=2), encoding="utf-8")

    # verify the saved model reloads and gives the same predictions
    reloaded = models.load_model(best_path)
    sample = splits["test"][0][:8].astype(np.float32)
    assert np.allclose(model.predict(sample, verbose=0), reloaded.predict(sample, verbose=0), atol=1e-5), \
        "Reloaded model output differs from the trained model!"
    log.info("Saved model verified: %s", best_path)

    log.info("Test accuracy: %.2f%% | AUC: %.4f", results["accuracy"] * 100, results["auc"])
    log.info("DONE. All files are in: %s", out.resolve())


if __name__ == "__main__":
    main()
```
