```python

# ============================================================
# DAY 6 - EXAMPLE 3
# CIFAR-10 CNN IMAGE CLASSIFICATION
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

from tensorflow.keras.datasets import cifar10

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout,
    RandomFlip,
    RandomRotation,
    RandomZoom
)


# ============================================================
# 1 - SETTINGS
# ============================================================

RANDOM_STATE = 42

EPOCHS = 15

BATCH_SIZE = 64


# ============================================================
# 2 - RANDOM SEED
# ============================================================

np.random.seed(RANDOM_STATE)

tf.random.set_seed(RANDOM_STATE)


# ============================================================
# 3 - CLASS NAMES
# ============================================================

class_names = [

    "airplane",

    "automobile",

    "bird",

    "cat",

    "deer",

    "dog",

    "frog",

    "horse",

    "ship",

    "truck"

]


# ============================================================
# 4 - LOAD CIFAR-10 DATASET
# ============================================================

print()
print("==========================================")
print("LOADING CIFAR-10")
print("==========================================")


(X_train, y_train), (X_test, y_test) = cifar10.load_data()


print()
print("Training Data Shape:")
print(X_train.shape)

print()
print("Training Labels Shape:")
print(y_train.shape)

print()
print("Testing Data Shape:")
print(X_test.shape)

print()
print("Testing Labels Shape:")
print(y_test.shape)


# ============================================================
# EXPECTED:
#
# X_train = (50000, 32, 32, 3)
# y_train = (50000, 1)
# X_test  = (10000, 32, 32, 3)
# y_test  = (10000, 1)
# ============================================================


# ============================================================
# 5 - DISPLAY FIRST 12 IMAGES
# ============================================================

plt.figure(
    figsize=(10, 6)
)


for i in range(12):

    plt.subplot(
        3,
        4,
        i + 1
    )


    plt.imshow(
        X_train[i]
    )


    label = int(
        y_train[i][0]
    )


    plt.title(
        class_names[label]
    )


    plt.axis("off")


plt.tight_layout()

plt.show()


# ============================================================
# 6 - NORMALIZE IMAGES
# ============================================================

X_train = X_train.astype(
    "float32"
) / 255.0


X_test = X_test.astype(
    "float32"
) / 255.0


print()
print("==========================================")
print("NORMALIZATION")
print("==========================================")

print(
    "Training Minimum:",
    X_train.min()
)

print(
    "Training Maximum:",
    X_train.max()
)


# ============================================================
# 7 - TRAIN / VALIDATION SPLIT
# ============================================================

X_train, X_val, y_train, y_val = train_test_split(

    X_train,

    y_train,

    test_size=0.10,

    random_state=RANDOM_STATE,

    stratify=y_train

)


print()
print("==========================================")
print("TRAIN / VALIDATION SPLIT")
print("==========================================")

print(
    "Training:",
    X_train.shape
)

print(
    "Validation:",
    X_val.shape
)

print(
    "Testing:",
    X_test.shape
)


# ============================================================
# 8 - BUILD BASIC CNN
# ============================================================

print()
print("==========================================")
print("BUILDING CNN")
print("==========================================")


model = Sequential([


    # --------------------------------------------------------
    # CONVOLUTION 1
    # --------------------------------------------------------

    Conv2D(

        32,

        (3, 3),

        activation="relu",

        input_shape=(32, 32, 3)

    ),


    # --------------------------------------------------------
    # MAX POOLING 1
    # --------------------------------------------------------

    MaxPooling2D(
        (2, 2)
    ),


    # --------------------------------------------------------
    # CONVOLUTION 2
    # --------------------------------------------------------

    Conv2D(

        64,

        (3, 3),

        activation="relu"

    ),


    # --------------------------------------------------------
    # MAX POOLING 2
    # --------------------------------------------------------

    MaxPooling2D(
        (2, 2)
    ),


    # --------------------------------------------------------
    # CONVOLUTION 3
    # --------------------------------------------------------

    Conv2D(

        128,

        (3, 3),

        activation="relu"

    ),


    # --------------------------------------------------------
    # MAX POOLING 3
    # --------------------------------------------------------

    MaxPooling2D(
        (2, 2)
    ),


    # --------------------------------------------------------
    # FLATTEN
    # --------------------------------------------------------

    Flatten(),


    # --------------------------------------------------------
    # DENSE
    # --------------------------------------------------------

    Dense(

        128,

        activation="relu"

    ),


    # --------------------------------------------------------
    # DROPOUT
    # --------------------------------------------------------

    Dropout(
        0.5
    ),


    # --------------------------------------------------------
    # OUTPUT
    # 10 CLASSES
    # --------------------------------------------------------

    Dense(

        10,

        activation="softmax"

    )

])


# ============================================================
# 9 - MODEL SUMMARY
# ============================================================

model.summary()


# ============================================================
# 10 - COMPILE MODEL
# ============================================================

model.compile(

    optimizer="adam",

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]

)


# ============================================================
# 11 - TRAIN BASIC CNN
# ============================================================

print()
print("==========================================")
print("START TRAINING")
print("==========================================")


history = model.fit(

    X_train,

    y_train,

    epochs=EPOCHS,

    batch_size=BATCH_SIZE,

    validation_data=(
        X_val,
        y_val
    )

)


# ============================================================
# 12 - TRAINING ACCURACY
# ============================================================

plt.figure(
    figsize=(8, 5)
)


plt.plot(

    history.history["accuracy"],

    label="Training Accuracy"

)


plt.plot(

    history.history["val_accuracy"],

    label="Validation Accuracy"

)


plt.title(
    "CIFAR-10 Training and Validation Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 13 - TRAINING LOSS
# ============================================================

plt.figure(
    figsize=(8, 5)
)


plt.plot(

    history.history["loss"],

    label="Training Loss"

)


plt.plot(

    history.history["val_loss"],

    label="Validation Loss"

)


plt.title(
    "CIFAR-10 Training and Validation Loss"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 14 - TEST MODEL
# ============================================================

print()
print("==========================================")
print("TESTING MODEL")
print("==========================================")


test_loss, test_accuracy = model.evaluate(

    X_test,

    y_test,

    verbose=1

)


print()
print("==========================================")
print("TEST RESULTS")
print("==========================================")

print(
    "Test Loss:",
    test_loss
)

print(
    "Test Accuracy:",
    test_accuracy
)

print(
    "Test Accuracy %:",
    test_accuracy * 100
)


# ============================================================
# 15 - PREDICTIONS
# ============================================================

print()
print("Generating predictions...")


predictions = model.predict(

    X_test,

    verbose=1

)


# Get class with highest probability

y_pred = predictions.argmax(
    axis=1
)


# Convert y_test from (10000, 1)
# to (10000,)

y_true = y_test.flatten()


# ============================================================
# 16 - ACCURACY SCORE
# ============================================================

accuracy = accuracy_score(

    y_true,

    y_pred

)


print()
print("==========================================")
print("ACCURACY SCORE")
print("==========================================")

print(
    "Accuracy:",
    accuracy
)

print(
    "Accuracy %:",
    accuracy * 100
)


# ============================================================
# 17 - CLASSIFICATION REPORT
# ============================================================

print()
print("==========================================")
print("CLASSIFICATION REPORT")
print("==========================================")


report = classification_report(

    y_true,

    y_pred,

    target_names=class_names,

    digits=4

)


print(report)


# ============================================================
# 18 - CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_true,

    y_pred

)


print()
print("==========================================")
print("CONFUSION MATRIX")
print("==========================================")

print(cm)


# ============================================================
# 19 - DISPLAY CONFUSION MATRIX
# ============================================================

plt.figure(
    figsize=(10, 8)
)


sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    xticklabels=class_names,

    yticklabels=class_names

)


plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)

plt.title(
    "CIFAR-10 Confusion Matrix"
)

plt.tight_layout()

plt.show()


# ============================================================
# 20 - FIND WRONG PREDICTIONS
# ============================================================

wrong = []


for i in range(
    len(y_test)
):

    if y_true[i] != y_pred[i]:

        wrong.append(i)


print()
print("==========================================")
print("WRONG PREDICTIONS")
print("==========================================")

print(
    "Number of Wrong Predictions:",
    len(wrong)
)


# ============================================================
# 21 - DISPLAY 12 WRONG PREDICTIONS
# ============================================================

plt.figure(
    figsize=(12, 8)
)


for j, i in enumerate(
    wrong[:12]
):


    plt.subplot(

        3,

        4,

        j + 1

    )


    plt.imshow(
        X_test[i]
    )


    true_class = class_names[
        y_true[i]
    ]


    predicted_class = class_names[
        y_pred[i]
    ]


    plt.title(

        "True: " +
        true_class +

        "\nPred: " +
        predicted_class

    )


    plt.axis("off")


plt.tight_layout()

plt.show()


# ============================================================
# 22 - DATA AUGMENTATION
# ============================================================

print()
print("==========================================")
print("DATA AUGMENTATION MODEL")
print("==========================================")


data_augmentation = Sequential([

    RandomFlip(
        "horizontal"
    ),

    RandomRotation(
        0.001
    ),

    RandomZoom(
        0.001
    )

])


# ============================================================
# 23 - BUILD CNN + DATA AUGMENTATION
# ============================================================

augmented_model = Sequential([


    # --------------------------------------------------------
    # DATA AUGMENTATION
    # --------------------------------------------------------

    data_augmentation,


    # --------------------------------------------------------
    # CONVOLUTION 1
    # --------------------------------------------------------

    Conv2D(

        32,

        (3, 3),

        activation="relu",

        input_shape=(32, 32, 3)

    ),

    MaxPooling2D(
        (2, 2)
    ),


    # --------------------------------------------------------
    # CONVOLUTION 2
    # --------------------------------------------------------

    Conv2D(

        64,

        (3, 3),

        activation="relu"

    ),

    MaxPooling2D(
        (2, 2)
    ),


    # --------------------------------------------------------
    # CONVOLUTION 3
    # --------------------------------------------------------

    Conv2D(

        128,

        (3, 3),

        activation="relu"

    ),

    MaxPooling2D(
        (2, 2)
    ),


    # --------------------------------------------------------
    # FLATTEN
    # --------------------------------------------------------

    Flatten(),


    # --------------------------------------------------------
    # DENSE
    # --------------------------------------------------------

    Dense(

        128,

        activation="relu"

    ),


    # --------------------------------------------------------
    # DROPOUT
    # --------------------------------------------------------

    Dropout(
        0.5
    ),


    # --------------------------------------------------------
    # OUTPUT
    # --------------------------------------------------------

    Dense(

        10,

        activation="softmax"

    )

])


# ============================================================
# 24 - COMPILE AUGMENTED MODEL
# ============================================================

augmented_model.compile(

    optimizer="adam",

    loss="sparse_categorical_crossentropy",

    metrics=["accuracy"]

)


# ============================================================
# 25 - TRAIN AUGMENTED MODEL
# ============================================================

print()
print("==========================================")
print("TRAINING CNN + DATA AUGMENTATION")
print("==========================================")


augmented_history = augmented_model.fit(

    X_train,

    y_train,

    epochs=EPOCHS,

    batch_size=BATCH_SIZE,

    validation_data=(
        X_val,
        y_val
    )

)


# ============================================================
# 26 - AUGMENTED MODEL ACCURACY
# ============================================================

plt.figure(
    figsize=(8, 5)
)


plt.plot(

    augmented_history.history["accuracy"],

    label="Training Accuracy"

)


plt.plot(

    augmented_history.history["val_accuracy"],

    label="Validation Accuracy"

)


plt.title(
    "CNN + Data Augmentation Accuracy"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Accuracy"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 27 - AUGMENTED MODEL LOSS
# ============================================================

plt.figure(
    figsize=(8, 5)
)


plt.plot(

    augmented_history.history["loss"],

    label="Training Loss"

)


plt.plot(

    augmented_history.history["val_loss"],

    label="Validation Loss"

)


plt.title(
    "CNN + Data Augmentation Loss"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.legend()

plt.grid()

plt.show()


# ============================================================
# 28 - EVALUATE AUGMENTED MODEL
# ============================================================

print()
print("==========================================")
print("EVALUATING AUGMENTED MODEL")
print("==========================================")


aug_test_loss, aug_test_accuracy = augmented_model.evaluate(

    X_test,

    y_test,

    verbose=1

)


print()
print("Augmented Model Test Loss:")

print(
    aug_test_loss
)


print()
print("Augmented Model Test Accuracy:")

print(
    aug_test_accuracy
)


print()
print("Augmented Model Test Accuracy %:")

print(
    aug_test_accuracy * 100
)


# ============================================================
# 29 - AUGMENTED MODEL PREDICTIONS
# ============================================================

aug_predictions = augmented_model.predict(

    X_test,

    verbose=1

)


aug_y_pred = aug_predictions.argmax(

    axis=1

)


# ============================================================
# 30 - AUGMENTED CONFUSION MATRIX
# ============================================================

aug_cm = confusion_matrix(

    y_true,

    aug_y_pred

)


print()
print("==========================================")
print("AUGMENTED MODEL CONFUSION MATRIX")
print("==========================================")


print(
    aug_cm
)


plt.figure(
    figsize=(10, 8)
)


sns.heatmap(

    aug_cm,

    annot=True,

    fmt="d",

    xticklabels=class_names,

    yticklabels=class_names

)


plt.xlabel(
    "Predicted"
)

plt.ylabel(
    "Actual"
)

plt.title(
    "CIFAR-10 - CNN + Data Augmentation"
)

plt.tight_layout()

plt.show()


# ============================================================
# 31 - SAVE BASIC MODEL
# ============================================================

model.save(
    "cifar10_cnn_model.keras"
)


# ============================================================
# 32 - SAVE AUGMENTED MODEL
# ============================================================

augmented_model.save(
    "cifar10_cnn_augmented_model.keras"
)


# ============================================================
# 33 - FINAL SUMMARY
# ============================================================

print()
print("==========================================")
print("FINAL SUMMARY")
print("==========================================")


print(
    "Training Images:",
    len(X_train)
)


print(
    "Validation Images:",
    len(X_val)
)


print(
    "Testing Images:",
    len(X_test)
)


print()
print(
    "Basic CNN Accuracy:",
    round(
        test_accuracy * 100,
        2
    ),
    "%"
)


print()
print(
    "CNN + Data Augmentation Accuracy:",
    round(
        aug_test_accuracy * 100,
        2
    ),
    "%"
)


print()
print("Saved Models:")


print(
    "cifar10_cnn_model.keras"
)


print(
    "cifar10_cnn_augmented_model.keras"
)


print()
print("==========================================")
print("CIFAR-10 PROJECT FINISHED")
print("==========================================")

```

```python
"""
Cats vs Dogs - CNN Classifier (Professional Version)
====================================================

Usage examples:
    python cats_vs_dogs_cnn.py
    python cats_vs_dogs_cnn.py --data "C:\\path\\to\\train" --epochs 30 --show
    python cats_vs_dogs_cnn.py --predict "C:\\path\\to\\image.jpg" --model outputs\\run_xxx\\best_model.keras
"""

from __future__ import annotations

import argparse
import json
import logging
import random
from concurrent.futures import ThreadPoolExecutor
from dataclasses import dataclass, asdict
from datetime import datetime
from pathlib import Path

import cv2
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
import tensorflow as tf
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    roc_curve,
)
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras import layers, models, regularizers
from tensorflow.keras.callbacks import (
    CSVLogger,
    EarlyStopping,
    ModelCheckpoint,
    ReduceLROnPlateau,
)

logging.basicConfig(level=logging.INFO, format="%(asctime)s | %(levelname)s | %(message)s")
log = logging.getLogger("cats_dogs")

CLASS_NAMES = ["Cat", "Dog"]  # label 0 = Cat, label 1 = Dog
VALID_EXT = (".jpg", ".jpeg", ".png")


# ============================================================
# 1 - CONFIGURATION
# ============================================================
@dataclass
class Config:
    data_dir: str = r"C:\Users\PC-LAB1\Desktop\New folder\day 5\dogs-vs-cats\train\train"
    output_dir: str = "outputs"
    img_size: int = 128
    per_class: int = 0              # 0 = use ALL images (balanced to the smaller class)
    test_size: float = 0.20
    val_size: float = 0.20          # fraction of the remaining train data
    batch_size: int = 64
    epochs: int = 5
    learning_rate: float = 1e-3
    resume_from: str = r"outputs\run_20260928_095655\best_model.keras"  # "" = train a NEW model
    resume_lr: float = 3e-4         # smaller LR when continuing an already-trained model
    dropout: float = 0.5
    l2: float = 1e-4
    seed: int = 42
    workers: int = 8
    show: bool = True               # show plots with plt.show() (they are always saved too)


def parse_args() -> tuple[Config, str | None, str | None]:
    d = Config()
    p = argparse.ArgumentParser(description="Cats vs Dogs CNN")
    p.add_argument("--data", dest="data_dir", default=d.data_dir)
    p.add_argument("--output", dest="output_dir", default=d.output_dir)
    p.add_argument("--img-size", type=int, default=d.img_size)
    p.add_argument("--per-class", type=int, default=d.per_class)
    p.add_argument("--batch-size", type=int, default=d.batch_size)
    p.add_argument("--epochs", type=int, default=d.epochs)
    p.add_argument("--lr", dest="learning_rate", type=float, default=d.learning_rate)
    p.add_argument("--seed", type=int, default=d.seed)
    p.add_argument("--resume", dest="resume_from", default=d.resume_from,
                   help="continue training this .keras model (empty string = new model)")
    p.add_argument("--no-show", action="store_true", help="do not display plots (only save them)")
    p.add_argument("--predict", default=None, help="path of a new image to classify")
    p.add_argument("--model", default=None, help="path of a saved .keras model (for --predict)")
    # parse_known_args: ignores extra args injected by Jupyter/IPython (e.g. --f=kernel.json)
    a, _ = p.parse_known_args()
    cfg = Config(
        data_dir=a.data_dir, output_dir=a.output_dir, img_size=a.img_size,
        per_class=a.per_class, batch_size=a.batch_size, epochs=a.epochs,
        learning_rate=a.learning_rate, seed=a.seed, show=not a.no_show,
        resume_from=a.resume_from,
    )
    return cfg, a.predict, a.model


def set_seed(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)
    tf.keras.utils.set_random_seed(seed)


# ============================================================
# 2 - DATA LOADING
# ============================================================
def load_image(path: Path, size: int) -> np.ndarray | None:
    """Read an image as RGB uint8 (size x size). Returns None if unreadable."""
    img = cv2.imread(str(path))
    if img is None:
        return None
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    return cv2.resize(img, (size, size), interpolation=cv2.INTER_AREA)


def list_files(data_dir: Path, prefix: str) -> list[Path]:
    return sorted(
        f for f in data_dir.iterdir()
        if f.name.lower().startswith(prefix) and f.suffix.lower() in VALID_EXT
    )


def select_files(files: list[Path], n: int, rng: random.Random) -> list[Path]:
    if len(files) < n:
        raise ValueError(f"Not enough images: found {len(files)}, need {n}")
    return rng.sample(files, n)


def load_dataset(cfg: Config) -> tuple[np.ndarray, np.ndarray, np.ndarray]:
    data_dir = Path(cfg.data_dir)
    if not data_dir.is_dir():
        raise FileNotFoundError(f"Dataset path not found: {data_dir}")

    rng = random.Random(cfg.seed)
    all_cats = list_files(data_dir, "cat.")
    all_dogs = list_files(data_dir, "dog.")
    log.info("Available: %d cats, %d dogs", len(all_cats), len(all_dogs))

    # per_class = 0  ->  use every image (balanced to the smaller class)
    n = cfg.per_class if cfg.per_class > 0 else min(len(all_cats), len(all_dogs))
    cats = select_files(all_cats, n, rng)
    dogs = select_files(all_dogs, n, rng)
    paths = cats + dogs
    labels = [0] * len(cats) + [1] * len(dogs)

    log.info("Loading %d images with %d threads...", len(paths), cfg.workers)
    with ThreadPoolExecutor(max_workers=cfg.workers) as pool:
        images = list(pool.map(lambda p: load_image(p, cfg.img_size), paths))

    keep = [i for i, im in enumerate(images) if im is not None]
    skipped = len(paths) - len(keep)
    if skipped:
        log.warning("Skipped %d unreadable images", skipped)

    X = np.stack([images[i] for i in keep]).astype(np.uint8)   # uint8 = 4x less RAM
    y = np.array([labels[i] for i in keep], dtype=np.int32)
    names = np.array([paths[i].name for i in keep])
    log.info("Loaded X=%s | cats=%d | dogs=%d", X.shape, (y == 0).sum(), (y == 1).sum())
    return X, y, names


def split_data(X, y, names, cfg: Config):
    """Stratified train / validation / test split."""
    idx = np.arange(len(y))
    idx_trainval, idx_test = train_test_split(
        idx, test_size=cfg.test_size, random_state=cfg.seed, stratify=y
    )
    idx_train, idx_val = train_test_split(
        idx_trainval, test_size=cfg.val_size, random_state=cfg.seed, stratify=y[idx_trainval]
    )
    parts = {"train": idx_train, "val": idx_val, "test": idx_test}
    for k, v in parts.items():
        log.info("%-5s: %4d images | cats=%d dogs=%d", k, len(v), (y[v] == 0).sum(), (y[v] == 1).sum())
    return {k: (X[v], y[v], names[v]) for k, v in parts.items()}


def make_dataset(X, y, cfg: Config, training: bool) -> tf.data.Dataset:
    ds = tf.data.Dataset.from_tensor_slices((X, y))
    if training:
        ds = ds.shuffle(len(y), seed=cfg.seed, reshuffle_each_iteration=True)
    ds = ds.batch(cfg.batch_size)
    ds = ds.map(lambda a, b: (tf.cast(a, tf.float32), tf.cast(b, tf.float32)),
                num_parallel_calls=tf.data.AUTOTUNE)
    return ds.prefetch(tf.data.AUTOTUNE)


# ============================================================
# 3 - MODEL
# ============================================================
def build_model(cfg: Config) -> tf.keras.Model:
    reg = regularizers.l2(cfg.l2)

    augmentation = tf.keras.Sequential(
        [layers.RandomFlip("horizontal"), layers.RandomRotation(0.10), layers.RandomZoom(0.10)],
        name="augmentation",
    )

    inputs = layers.Input(shape=(cfg.img_size, cfg.img_size, 3))
    x = layers.Rescaling(1.0 / 255)(inputs)
    x = augmentation(x)                       # active only during training

    for filters in (32, 64, 128, 256):
        x = layers.Conv2D(filters, 3, padding="same", use_bias=False, kernel_regularizer=reg)(x)
        x = layers.BatchNormalization()(x)
        x = layers.ReLU()(x)
        x = layers.MaxPooling2D()(x)

    x = layers.GlobalAveragePooling2D()(x)    # far fewer parameters than Flatten
    x = layers.Dense(128, activation="relu", kernel_regularizer=reg)(x)
    x = layers.Dropout(cfg.dropout)(x)
    outputs = layers.Dense(1, activation="sigmoid")(x)   # Cat=0, Dog=1

    model = models.Model(inputs, outputs, name="cats_vs_dogs_cnn")
    compile_model(model, cfg.learning_rate)
    return model


def compile_model(model: tf.keras.Model, lr: float) -> None:
    model.compile(
        optimizer=tf.keras.optimizers.Adam(lr),
        loss="binary_crossentropy",
        metrics=[
            "accuracy",
            tf.keras.metrics.AUC(name="auc"),
            tf.keras.metrics.Precision(name="precision"),
            tf.keras.metrics.Recall(name="recall"),
        ],
    )


def get_model(cfg: Config) -> tf.keras.Model:
    """Load the previous model and continue training it, or build a new one."""
    if cfg.resume_from:
        path = Path(cfg.resume_from)
        if not path.is_file():
            raise FileNotFoundError(f"Model to resume not found: {path.resolve()}")
        log.info("Resuming training from: %s", path)
        model = models.load_model(path)
        compile_model(model, cfg.resume_lr)   # fresh optimizer, smaller learning rate
        return model
    log.info("Building a NEW model")
    return build_model(cfg)


# ============================================================
# 4 - PLOTS (always saved, optionally shown)
# ============================================================
sns.set_theme(style="whitegrid", context="notebook")
plt.rcParams.update({
    "figure.dpi": 100,
    "axes.titleweight": "bold",
    "axes.titlesize": 12,
    "axes.spines.top": False,
    "axes.spines.right": False,
})
COLORS = {"Cat": "#4C72B0", "Dog": "#DD8452", "good": "#2E9E5B", "bad": "#D64545"}


def finish_fig(fig, path: Path, show: bool) -> None:
    fig.tight_layout()
    fig.savefig(path, dpi=150, bbox_inches="tight")
    if show:
        plt.show()
    plt.close(fig)


def plot_class_distribution(y_train, y_val, y_test, out: Path, show: bool) -> None:
    """Bar chart of cats/dogs count in each split."""
    splits = {"Train": y_train, "Validation": y_val, "Test": y_test}
    x = np.arange(len(splits))
    w = 0.38
    fig, ax = plt.subplots(figsize=(7, 4.5))
    for k, cls in enumerate(CLASS_NAMES):
        counts = [int((v == k).sum()) for v in splits.values()]
        bars = ax.bar(x + (k - 0.5) * w, counts, w, label=cls, color=COLORS[cls])
        ax.bar_label(bars, padding=2, fontsize=9)
    ax.set_xticks(x, list(splits))
    ax.set(title="Class distribution per split", ylabel="Images")
    ax.legend()
    finish_fig(fig, out / "class_distribution.png", show)


def plot_history(history: dict, out: Path, show: bool) -> None:
    """Accuracy / Loss (with best epoch marked) and learning rate if available."""
    best = int(np.argmin(history["val_loss"]))
    lr = history.get("learning_rate") or history.get("lr")
    ncols = 3 if lr else 2
    fig, axes = plt.subplots(1, ncols, figsize=(6 * ncols, 4.5))

    for ax, metric, title in zip(axes[:2], ("accuracy", "loss"), ("Accuracy", "Loss")):
        ax.plot(history[metric], label="Train", lw=2)
        ax.plot(history[f"val_{metric}"], label="Validation", lw=2)
        ax.axvline(best, color="gray", ls="--", lw=1)
        ax.scatter(best, history[f"val_{metric}"][best], color=COLORS["bad"], zorder=5,
                   label=f"Best epoch = {best + 1}")
        ax.set(title=title, xlabel="Epoch", ylabel=title)
        ax.legend()
    if lr:
        axes[2].plot(lr, lw=2, color="#8172B3")
        axes[2].set(title="Learning rate", xlabel="Epoch", ylabel="LR", yscale="log")
    fig.suptitle("Training history", fontweight="bold", y=1.02)
    finish_fig(fig, out / "training_curves.png", show)


def plot_confusion(cm, out: Path, show: bool) -> None:
    """Confusion matrix with counts and row percentages."""
    pct = cm / cm.sum(axis=1, keepdims=True) * 100
    labels = np.array([[f"{cm[i, j]}\n({pct[i, j]:.1f}%)" for j in range(2)] for i in range(2)])
    fig, ax = plt.subplots(figsize=(6, 5))
    sns.heatmap(cm, annot=labels, fmt="", cmap="Blues", cbar=False, linewidths=2,
                linecolor="white", xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES,
                annot_kws={"fontsize": 13}, ax=ax)
    ax.set(xlabel="Predicted", ylabel="Actual",
           title=f"Confusion Matrix (accuracy {np.trace(cm) / cm.sum() * 100:.2f}%)")
    finish_fig(fig, out / "confusion_matrix.png", show)


def plot_roc(y_true, probs, out: Path, show: bool) -> None:
    fpr, tpr, _ = roc_curve(y_true, probs)
    auc = roc_auc_score(y_true, probs)
    fig, ax = plt.subplots(figsize=(6, 5.5))
    ax.plot(fpr, tpr, lw=2.5, label=f"CNN (AUC = {auc:.4f})")
    ax.fill_between(fpr, tpr, alpha=0.15)
    ax.plot([0, 1], [0, 1], "--", color="gray", label="Random")
    ax.set(xlabel="False Positive Rate", ylabel="True Positive Rate", title="ROC Curve",
           xlim=(0, 1), ylim=(0, 1.02))
    ax.legend(loc="lower right")
    finish_fig(fig, out / "roc_curve.png", show)


def plot_probability_hist(y_true, probs, out: Path, show: bool) -> None:
    """How confident is the model? Ideal: cats near 0, dogs near 1."""
    fig, ax = plt.subplots(figsize=(7.5, 4.5))
    for k, cls in enumerate(CLASS_NAMES):
        ax.hist(probs[y_true == k], bins=30, range=(0, 1), alpha=0.65,
                label=f"Actual {cls}", color=COLORS[cls])
    ax.axvline(0.5, color="black", ls="--", lw=1, label="Threshold 0.5")
    ax.set(title="Predicted probability of Dog", xlabel="P(Dog)", ylabel="Images")
    ax.legend()
    finish_fig(fig, out / "probability_histogram.png", show)


def plot_samples(X, y_true, y_pred, probs, indices, title, path: Path, show: bool,
                 cols: int = 5) -> None:
    """Grid of images; colored border/title: green = correct, red = wrong."""
    indices = list(indices)[:10]
    if not indices:
        return
    rows = int(np.ceil(len(indices) / cols))
    fig, axes = plt.subplots(rows, cols, figsize=(3.2 * cols, 3.5 * rows), squeeze=False)
    for ax in axes.ravel():
        ax.axis("off")
    for ax, i in zip(axes.ravel(), indices):
        ok = y_true[i] == y_pred[i]
        color = COLORS["good"] if ok else COLORS["bad"]
        conf = probs[i] if y_pred[i] == 1 else 1 - probs[i]
        ax.imshow(X[i])
        ax.set_title(f"Actual: {CLASS_NAMES[y_true[i]]}\nPred: {CLASS_NAMES[y_pred[i]]} "
                     f"({conf * 100:.1f}%)", fontsize=10, color=color)
        for s in ax.spines.values():
            s.set_visible(True)
            s.set_edgecolor(color)
            s.set_linewidth(3)
    fig.suptitle(title, fontsize=14, fontweight="bold")
    finish_fig(fig, path, show)


# ============================================================
# 5 - EVALUATION
# ============================================================
def evaluate(model, test_data, cfg: Config, out: Path) -> dict:
    X_test, y_test, names_test = test_data
    test_ds = make_dataset(X_test, y_test, cfg, training=False)

    metrics = model.evaluate(test_ds, return_dict=True, verbose=0)
    probs = model.predict(test_ds, verbose=0).ravel()
    y_pred = (probs >= 0.5).astype(int)

    cm = confusion_matrix(y_test, y_pred)
    report = classification_report(y_test, y_pred, target_names=CLASS_NAMES, digits=4)
    log.info("Test metrics: %s", {k: round(float(v), 4) for k, v in metrics.items()})
    print("\n" + report)
    print("Confusion matrix:\n", cm)

    plot_confusion(cm, out, cfg.show)
    plot_roc(y_test, probs, out, cfg.show)
    plot_probability_hist(y_test, probs, out, cfg.show)
    plot_samples(X_test, y_test, y_pred, probs, range(len(y_test)),
                 "Random test predictions", out / "test_samples.png", cfg.show)
    wrong = np.where(y_pred != y_test)[0]
    plot_samples(X_test, y_test, y_pred, probs, wrong,
                 f"Misclassified images ({len(wrong)} total)", out / "misclassified.png", cfg.show)

    pd.DataFrame({
        "file": names_test,
        "actual": [CLASS_NAMES[i] for i in y_test],
        "predicted": [CLASS_NAMES[i] for i in y_pred],
        "probability_dog": probs,
        "confidence": np.where(y_pred == 1, probs, 1 - probs),
        "correct": y_pred == y_test,
    }).to_csv(out / "test_predictions.csv", index=False)

    (out / "test_report.txt").write_text(report, encoding="utf-8")
    return {k: float(v) for k, v in metrics.items()}


# ============================================================
# 6 - INFERENCE ON A NEW IMAGE
# ============================================================
def predict_image(model: tf.keras.Model, image_path: str, img_size: int, show: bool = True) -> None:
    path = Path(image_path)
    if not path.is_file():
        log.error("Image not found: %s", path)
        return
    img = load_image(path, img_size)
    if img is None:
        log.error("Could not read image: %s", path)
        return

    prob = float(model.predict(img[np.newaxis].astype(np.float32), verbose=0)[0][0])
    label = CLASS_NAMES[int(prob >= 0.5)]
    conf = (prob if prob >= 0.5 else 1 - prob) * 100
    log.info("Prediction: %s (%.2f%% confidence)", label, conf)

    if show:
        plt.imshow(img)
        plt.title(f"{label} - {conf:.2f}%")
        plt.axis("off")
        plt.show()


# ============================================================
# 7 - MAIN
# ============================================================
def main() -> None:
    cfg, predict_path, model_path = parse_args()
    set_seed(cfg.seed)

    # ---- inference-only mode ----
    if predict_path:
        if not model_path:
            raise SystemExit("--predict requires --model <path to .keras file>")
        predict_image(models.load_model(model_path), predict_path, cfg.img_size)
        return

    # ---- training mode ----
    out = Path(cfg.output_dir) / f"run_{datetime.now():%Y%m%d_%H%M%S}"
    out.mkdir(parents=True, exist_ok=True)
    (out / "config.json").write_text(json.dumps(asdict(cfg), indent=2), encoding="utf-8")
    log.info("Outputs will be saved in: %s", out.resolve())

    X, y, names = load_dataset(cfg)
    splits = split_data(X, y, names, cfg)

    plot_class_distribution(splits["train"][1], splits["val"][1], splits["test"][1], out, cfg.show)
    plot_samples(X, y, y, np.where(y == 1, 1.0, 0.0), np.random.permutation(len(y))[:10],
                 "Random training samples", out / "dataset_samples.png", cfg.show)

    train_ds = make_dataset(*splits["train"][:2], cfg, training=True)
    val_ds = make_dataset(*splits["val"][:2], cfg, training=False)

    weights = compute_class_weight("balanced", classes=np.array([0, 1]), y=splits["train"][1])
    class_weight = {0: float(weights[0]), 1: float(weights[1])}
    log.info("Class weights: %s", class_weight)

    model = get_model(cfg)
    model.summary()

    best_path = out / "best_model.keras"
    callbacks = [
        EarlyStopping(monitor="val_loss", patience=8, restore_best_weights=True, verbose=1),
        ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, min_lr=1e-6, verbose=1),
        ModelCheckpoint(str(best_path), monitor="val_loss", save_best_only=True),
        CSVLogger(str(out / "training_log.csv")),
    ]

    history = model.fit(
        train_ds, validation_data=val_ds, epochs=cfg.epochs,
        class_weight=class_weight, callbacks=callbacks, verbose=1,
    )
    plot_history(history.history, out, cfg.show)

    results = evaluate(model, splits["test"], cfg, out)
    (out / "test_metrics.json").write_text(json.dumps(results, indent=2), encoding="utf-8")

    # verify the saved model reloads and gives the same predictions
    reloaded = models.load_model(best_path)
    sample = splits["test"][0][:8].astype(np.float32)
    assert np.allclose(model.predict(sample, verbose=0), reloaded.predict(sample, verbose=0), atol=1e-5), \
        "Reloaded model output differs from the trained model!"
    log.info("Saved model verified: %s", best_path)

    log.info("Test accuracy: %.2f%% | AUC: %.4f", results["accuracy"] * 100, results["auc"])
    log.info("DONE. All files are in: %s", out.resolve())


if __name__ == "__main__":
    main()
```
