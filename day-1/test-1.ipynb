```python
import numpy as np
import matplotlib.pyplot as plt

image = np.zeros((400, 400), dtype=np.uint8)

image[100:300, 100:300] = 255

image[130:270, 130:140] = 0
image[130:140, 130:180] = 0
image[130:270, 170:180] = 0
image[130:140, 170:220] = 0

image[130:270, 220:230] = 0

pixels = image.shape[0] * image.shape[1]
print("Number of pixels:", pixels)

plt.imshow(image, cmap="gray")
plt.axis("off")
plt.show()
```

```python
from PIL import Image, ImageEnhance
import matplotlib.pyplot as plt

# قراءة الصورة
img = Image.open(r"C:\Users\PC-LAB1\Downloads\output.png")

# تحويل الصورة إلى Grayscale
gray = img.convert("L")

# تحسين التباين
enhancer = ImageEnhance.Contrast(gray)
contrast_img = enhancer.enhance(2.0)

# عرض الصورة الأصلية والمعدلة
plt.figure(figsize=(12, 6))

plt.subplot(2, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(contrast_img, cmap="gray")
plt.title("Contrast Enhanced")
plt.axis("off")

# Histogram للصورة الأصلية
plt.subplot(2, 2, 3)
plt.hist(gray.getdata(), bins=256, range=(0, 256))
plt.title("Original Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

# Histogram بعد تحسين التباين
plt.subplot(2, 2, 4)
plt.hist(contrast_img.getdata(), bins=256, range=(0, 256))
plt.title("Enhanced Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```
