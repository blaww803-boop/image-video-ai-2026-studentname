```python
import cv2

rtsp_url = "rtsp://192.168.1.100:8554/mystream"

cap = cv2.VideoCapture(rtsp_url)

if not cap.isOpened():
    print("Failed to open RTSP stream")
    exit()

while True:
    ret, frame = cap.read()

    if not ret:
        print("Failed to read frame")
        break

    cv2.imshow("RTSP Stream", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
from ultralytics import YOLO

# ============================================================
# 1 - RTSP STREAM
# ============================================================

rtsp_url = "rtsp://192.168.1.100:8554/mystream"

# ============================================================
# 2 - LOAD YOLO
# ============================================================

model = YOLO("yolo11n.pt")

# ============================================================
# 3 - OPEN RTSP
# ============================================================

cap = cv2.VideoCapture(rtsp_url)

if not cap.isOpened():
    print("Failed to open RTSP stream")
    exit()

print("RTSP connected successfully")

# ============================================================
# 4 - READ FRAMES
# ============================================================

while True:

    ret, frame = cap.read()

    if not ret:
        print("Failed to read frame")
        break

    # ========================================================
    # YOLO TRACKING
    # 0 = person
    # 2 = car
    # ========================================================

    results = model.track(
        frame,
        persist=True,
        tracker="bytetrack.yaml",
        classes=[0, 2],
        conf=0.35,
        verbose=False
    )

    # ========================================================
    # DRAW RESULTS
    # ========================================================

    annotated_frame = results[0].plot()

    cv2.imshow("YOLO RTSP Tracking", annotated_frame)

    # Q = Exit
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

# ============================================================
# 5 - RELEASE
# ============================================================

cap.release()
cv2.destroyAllWindows()

print("Program finished")
```
