```python
import cv2
import matplotlib.pyplot as plt

# قراءة الصورة
image = cv2.imread(r"C:\Users\PC-LAB1\Desktop\New folder\WhatsApp Image 2026-09-22 at 9.04.15 AM (1).jpeg")

# تحويل إلى Gray
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Threshold
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# إنشاء Kernel
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

# Erosion
eroded = cv2.erode(binary, kernel, iterations=1)

# Dilation
dilated = cv2.dilate(binary, kernel, iterations=1)

# Opening
opened = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)

# Closing
closed = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

# عرض النتائج
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(binary, cmap="gray")
plt.title("Binary")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(eroded, cmap="gray")
plt.title("Erosion")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(dilated, cmap="gray")
plt.title("Dilation")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(opened, cmap="gray")
plt.title("Opening")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(closed, cmap="gray")
plt.title("Closing")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# قراءة الصورة
image = cv2.imread(r"C:\Users\PC-LAB1\Desktop\New folder\WhatsApp Image 2026-09-22 at 9.04.15 AM (1).jpeg")

# تحويل إلى Gray
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Threshold
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# اكتشاف Contours
contours, hierarchy = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# نسخ الصورة
result = image.copy()

# استخراج الخصائص
for i, contour in enumerate(contours):

    # حساب Area
    area = cv2.contourArea(contour)

    # Bounding Box
    x, y, w, h = cv2.boundingRect(contour)

    # رسم Bounding Box
    cv2.rectangle(result, (x, y), (x + w, y + h), (0, 255, 0), 2)

    # طباعة الخصائص
    print("Object ID:", i + 1)
    print("Area:", area)
    print("X:", x)
    print("Y:", y)
    print("Width:", w)
    print("Height:", h)
    print("--------------------")

# تحويل إلى RGB
result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)

# عرض النتائج
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(binary, cmap="gray")
plt.title("Binary")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(result_rgb)
plt.title("Contours")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# قراءة الصورة
image = cv2.imread(r"C:\Users\PC-LAB1\Desktop\New folder\WhatsApp Image 2026-09-22 at 9.04.15 AM (1).jpeg")

# تحويل إلى Gray
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Gaussian Blur
blur = cv2.GaussianBlur(gray, (5, 5), 0)

# Threshold
_, binary = cv2.threshold(blur, 127, 255, cv2.THRESH_BINARY)

# إنشاء Kernel
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

# Morphology
clean = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)

# Edge Detection
edges = cv2.Canny(clean, 50, 150)

# اكتشاف Contours
contours, _ = cv2.findContours(clean, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# نسخ الصورة
result = image.copy()

# رسم الأجسام
for contour in contours:

    # حساب Area
    area = cv2.contourArea(contour)

    # تجاهل الأجسام الصغيرة
    if area > 100:

        # Bounding Box
        x, y, w, h = cv2.boundingRect(contour)

        # رسم المستطيل
        cv2.rectangle(result, (x, y), (x + w, y + h), (0, 255, 0), 2)

# تحويل إلى RGB
result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)

# عرض النتائج
plt.figure(figsize=(15, 8))

plt.subplot(2, 3, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(gray, cmap="gray")
plt.title("Gray")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(binary, cmap="gray")
plt.title("Binary")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(clean, cmap="gray")
plt.title("Morphology")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(edges, cmap="gray")
plt.title("Edges")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(result_rgb)
plt.title("Features")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# ==============================
# 1 - قراءة الصورة
# ==============================

image = cv2.imread(r"C:\Users\PC-LAB1\Desktop\New folder\WhatsApp Image 2026-09-22 at 9.04.15 AM (1).jpeg")

# إنشاء مجلد لحفظ النتائج
output_dir = "results"
os.makedirs(output_dir, exist_ok=True)

# ==============================
# 2 - معلومات الصورة
# ==============================

height, width, channels = image.shape
data_type = image.dtype

print("Image Information")
print("------------------")
print("Width:", width)
print("Height:", height)
print("Channels:", channels)
print("Data Type:", data_type)

# ==============================
# 3 - Grayscale
# ==============================

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

cv2.imwrite(os.path.join(output_dir, "grayscale.jpg"), gray)

# ==============================
# 4 - Binary
# ==============================

_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

cv2.imwrite(os.path.join(output_dir, "binary.jpg"), binary)

# ==============================
# 5 - Histogram
# ==============================

hist = cv2.calcHist([gray], [0], None, [256], [0, 256])

plt.figure()
plt.plot(hist)
plt.title("Grayscale Histogram")
plt.xlabel("Intensity")
plt.ylabel("Frequency")
plt.savefig(os.path.join(output_dir, "histogram.png"))
plt.show()

# ==============================
# 6 - Morphology
# ==============================

kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

morphology = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)

cv2.imwrite(os.path.join(output_dir, "morphology.jpg"), morphology)

# ==============================
# 7 - Edge Detection
# ==============================

edges = cv2.Canny(gray, 50, 150)

cv2.imwrite(os.path.join(output_dir, "edges.jpg"), edges)

# ==============================
# 8 - Contours
# ==============================

contours, hierarchy = cv2.findContours(
    morphology,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

result = image.copy()

# ملف التقرير
report_path = os.path.join(output_dir, "report.txt")

with open(report_path, "w") as report:

    report.write("IMAGE ANALYSIS REPORT\n")
    report.write("====================\n\n")

    report.write("Image Information\n")
    report.write("-----------------\n")
    report.write(f"Width: {width}\n")
    report.write(f"Height: {height}\n")
    report.write(f"Channels: {channels}\n")
    report.write(f"Data Type: {data_type}\n\n")

    report.write("Detected Objects\n")
    report.write("----------------\n")

    object_id = 0

    for contour in contours:

        area = cv2.contourArea(contour)

        # تجاهل الأجسام الصغيرة
        if area > 100:

            object_id += 1

            x, y, w, h = cv2.boundingRect(contour)

            # رسم Bounding Box
            cv2.rectangle(
                result,
                (x, y),
                (x + w, y + h),
                (0, 255, 0),
                2
            )

            # كتابة رقم الجسم
            cv2.putText(
                result,
                f"ID: {object_id}",
                (x, y - 10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (0, 255, 0),
                2
            )

            # كتابة البيانات في التقرير
            report.write(f"Object ID: {object_id}\n")
            report.write(f"Area: {area}\n")
            report.write(f"Bounding Box X: {x}\n")
            report.write(f"Bounding Box Y: {y}\n")
            report.write(f"Width: {w}\n")
            report.write(f"Height: {h}\n")
            report.write("-----------------\n")

    report.write(f"\nTotal Objects: {object_id}\n")

# ==============================
# 9 - حفظ الصورة النهائية
# ==============================

cv2.imwrite(
    os.path.join(output_dir, "final_result.jpg"),
    result
)

# ==============================
# 10 - عرض النتائج باستخدام plt
# ==============================

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(image_rgb)
plt.title("Original")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(binary, cmap="gray")
plt.title("Binary")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(morphology, cmap="gray")
plt.title("Morphology")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(edges, cmap="gray")
plt.title("Edges")
plt.axis("off")

plt.subplot(2, 3, 6)
plt.imshow(result_rgb)
plt.title("Final Result")
plt.axis("off")

plt.tight_layout()
plt.show()

print("\nAnalysis completed successfully!")
print("Results saved in:", output_dir)
```
